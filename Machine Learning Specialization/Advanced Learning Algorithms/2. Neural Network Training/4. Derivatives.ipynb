```python
import sympy

# Define symbolic variables
J, w = sympy.symbols('J,w')

# Example 1: Quadratic function (J = w^2)
J = w**2
dJ_dw = sympy.diff(J, w)
print(dJ_dw)
print(dJ_dw.subs([(w, 2)]))

# Example 2: Cubic function (J = w^3)
J = w**3
dJ_dw = sympy.diff(J, w)
print(dJ_dw)
print(dJ_dw.subs([(w, 2)]))

# Example 3: Reciprocal function (J = 1/w)
J = 1/w
dJ_dw = sympy.diff(J, w)
print(dJ_dw)
print(dJ_dw.subs([(w, 2)]))
```
