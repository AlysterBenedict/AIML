```python
# Cell 1: Forward & Backpropagation from Scratch (NumPy)
import numpy as np

# 1. Generate synthetic dataset for binary classification
np.random.seed(42)
m = 200
n_features = 2

# Synthetic data generation
X = np.random.randn(m, n_features)
w_true = np.array([[2.0], [-3.0]])
b_true = 0.5
z_true = np.dot(X, w_true) + b_true
probs = 1 / (1 + np.exp(-z_true))
y = (probs >= 0.5).astype(float)

# Sigmoid Activation Function: g(z) = 1 / (1 + e^(-z))
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

# Binary Cross-Entropy Cost Function: J(w,b) = -1/m * sum(y*log(a) + (1-y)*log(1-a))
def compute_cost(a, y):
    m = y.shape[0]
    eps = 1e-15  # Numerical stability offset
    a_clipped = np.clip(a, eps, 1.0 - eps)
    cost = - (1.0 / m) * np.sum(y * np.log(a_clipped) + (1.0 - y) * np.log(1.0 - a_clipped))
    return cost

# Initialize parameters (Weights and Bias) to zero
w = np.zeros((n_features, 1))
b = 0.0

learning_rate = 0.1
epochs = 500

print("--- Training From Scratch (NumPy) ---")
for epoch in range(epochs):
    # 1. Forward Propagation: z = X*w + b, a = sigmoid(z)
    z = np.dot(X, w) + b
    a = sigmoid(z)
    
    # 2. Compute Loss & Cost
    cost = compute_cost(a, y)
    
    # 3. Backpropagation: Derivatives wrt z, w, and b
    # dz = dJ/dz = a - y
    dz = a - y
    dw = (1.0 / m) * np.dot(X.T, dz)
    db = (1.0 / m) * np.sum(dz)
    
    # 4. Gradient Descent Parameter Updates: w = w - alpha*dw, b = b - alpha*db
    w = w - learning_rate * dw
    b = b - learning_rate * db
    
    if (epoch + 1) % 100 == 0 or epoch == 0:
        print(f"Epoch {epoch+1:3d}/{epochs} - Cost: {cost:.4f}")

print("\nFinal Learned Parameters (Scratch NumPy):")
print(f"w: {w.ravel()}")
print(f"b: {b:.4f}")
#Learned all the parameters
```

```python
# Cell 2: Equivalent TensorFlow/Keras Implementation
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.losses import BinaryCrossentropy
from tensorflow.keras.optimizers import SGD

# 1. Define Model Architecture using tf.keras.Sequential
model = Sequential([
    Dense(units=1, activation='sigmoid', input_shape=(n_features,))
])

# Initialize weights to zeros to match scratch baseline initialization
model.layers[0].set_weights([np.zeros((n_features, 1)), np.array([0.0])])

# 2. Compile model with BinaryCrossentropy loss and SGD optimizer
optimizer = SGD(learning_rate=learning_rate) 
model.compile(optimizer=optimizer, loss=BinaryCrossentropy())

# 3. Train model on the synthetic dataset using model.fit()
print("--- Training Equivalent Model (TensorFlow) ---")
history = model.fit(X, y, epochs=epochs, batch_size=m, verbose=0)

# Print cost at interval epochs
for ep in range(0, epochs, 100):
    print(f"Epoch {ep+1:3d}/{epochs} - Cost: {history.history['loss'][ep]:.4f}")
print(f"Epoch {epochs:3d}/{epochs} - Cost: {history.history['loss'][-1]:.4f}")

w_tf, b_tf = model.layers[0].get_weights()
print("\nFinal Learned Parameters (TensorFlow):")
print(f"w: {w_tf.ravel()}")
print(f"b: {b_tf[0]:.4f}")
```
