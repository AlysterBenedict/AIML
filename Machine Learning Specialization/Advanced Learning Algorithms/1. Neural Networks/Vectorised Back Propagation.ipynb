```python
import numpy as np

# Define input matrix A (shape 2x3) and transpose AT (shape 3x2)
A = np.array([[1, -1, 0.1], [2, -2, 0.2]])
AT = np.array([[1, 2], [-1, -2], [0.1, 0.2]])

# Alternative NumPy transpose syntax:
AT = A.T

# Define weight matrix W (shape 2x4)
W = np.array([[3, 5, 7, 9], [4, 6,8,0]])

# Method 1: Using np.matmul
Z = np.matmul(AT, W)

# Method 2: Using the @ operator
Z = AT @ W
```
