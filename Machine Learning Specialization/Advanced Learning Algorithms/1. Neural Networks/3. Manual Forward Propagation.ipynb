```python
import numpy as np

def sigmoid(z):
    """
    Sigmoid activation function: g(z) = 1 / (1 + e^(-z))
    """
    return 1 / (1 + np.exp(-z))


def dense(a_in, W, b):
    """
    Computes activation output for a single dense (fully connected) layer.
    
    Parameters:
        a_in (ndarray): Input vector of shape (n_in,)
        W    (ndarray): Weight matrix of shape (n_in, units)
        b    (ndarray): Bias vector of shape (units,)
        
    Returns:
        a_out (ndarray): Layer activation vector of shape (units,)
    """
    units = W.shape[1]               # Number of neurons in the current layer
    a_out = np.zeros(units)          # Initialize output vector
    
    for j in range(units):
        w = W[:, j]                  # Extract j-th column weight vector
        z = np.dot(w, a_in) + b[j]   # Linear dot product + bias
        a_out[j] = sigmoid(z)        # Apply activation function
    return a_out


def sequential(x, weights_and_biases):
    """
    Chains multiple dense layers to perform forward propagation across a deep network.
    
    Parameters:
        x (ndarray): Network input feature vector
        weights_and_biases (list of tuples): List containing (W, b) for each layer
        
    Returns:
        f_x (ndarray): Final network output prediction
    """
    a = x
    for W, b in weights_and_biases:
        a = dense(a, W, b)
    f_x = a
    return f_x


# ---------------------------------------------------------
# Example Execution using parameters from the slide
# ---------------------------------------------------------

if __name__ == "__main__":
    # Input feature vector x (a_in)
    x = np.array([-2, 4])

    # Layer 1 parameters (2 input features, 3 hidden units)
    W1 = np.array([
        [1, -3,  5],
        [2,  4, -6]
    ])
    b1 = np.array([-1, 1, 2])

    # Test single dense layer computation
    a1 = dense(x, W1, b1)
    print("Layer 1 Activation Output (a1):", a1)

    # Example setup for a sequential multi-layer pass
    # Layer 2: 3 input activations -> 2 hidden units
    W2 = np.array([
        [1, -2],
        [3, -4],
        [5, -6]
    ])
    b2 = np.array([1, -1])

    # Layer 3: 2 input activations -> 1 output unit
    W3 = np.array([
        [0.5],
        [-0.5]
    ])
    b3 = np.array([0.1])

    layers = [(W1, b1), (W2, b2), (W3, b3)]
    final_output = sequential(x, layers)
    print("Sequential Network Output:", final_output)
```
